```python
import requests
from pathlib import Path

ARTICLE_ID = 21893808
VERSION = 3

url = f"https://api.figshare.com/v2/articles/{ARTICLE_ID}/versions/{VERSION}"

print("Recupero informazioni EWELD...")

r = requests.get(url)
r.raise_for_status()

data = r.json()

print("\nDataset:", data.get("title"))
print("File disponibili:")

for f in data["files"]:
    print(f"  - {f['name']} ({f['size'] / 1024 / 1024:.2f} MB)")

output_dir = Path("EWELD")
output_dir.mkdir(exist_ok=True)

print("\nInizio download...\n")

for f in data["files"]:
    filename = output_dir / f["name"]
    download_url = f["download_url"]

    print(f"Scarico: {f['name']}")

    with requests.get(download_url, stream=True) as response:
        response.raise_for_status()

        with open(filename, "wb") as out:
            for chunk in response.iter_content(chunk_size=1024 * 1024):
                if chunk:
                    out.write(chunk)

    print(f"  OK -> {filename}")

print("\nDownload completato!")
```
